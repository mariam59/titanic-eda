## Summary
- Out of 891 passengers, only 38% survived (342 survived) and 62% died (549 died).
- Sex was the strongest factor : about 74% of women survived (314 women) vs. about 18% of men (577 men).
- Ticket class mattered: first class had the highest survival rate (62%) and third class the lowest (24%).
- Age had a weak effect: the average age of survivors was 28.3 vs. 30.6 for those who died.
- Limitation: 177 passengers have a missing Age, so the child analysis is approximate.
